```python
def tokenize_and_align_labels(tokens, labels):
    tokenized_inputs = tokenizer(
        tokens,
        is_split_into_words=True,
        padding="max_length",
        truncation=True
    )

    word_ids = tokenized_inputs.word_ids()

    aligned_labels = []
    previous_word_id = None

    for word_id in word_ids:

        if word_id is None:
            aligned_labels.append(-100)
        elif word_id != previous_word_id:
            label = labels[word_id]
            aligned_labels.append(label2id[label])
        else:
            aligned_labels.append(-100)

        previous_word_id = word_id
    tokenized_inputs["labels"] = aligned_labels


    return tokenized_inputs
```

```python
def tokenize_and_align_labels_with_chunks(tokens, labels):
    tokenized_inputs = tokenizer(
        tokens,
        is_split_into_words=True,
        truncation=True,
        max_length=512,
        stride=128,
        return_overflowing_tokens=True
    )

    all_labels = []

    for chunk_index in range(len(tokenized_inputs["input_ids"])):

        word_ids = tokenized_inputs.word_ids(
            batch_index=chunk_index
        )

        aligned_labels = []
        previous_word_id = None

        for word_id in word_ids:

            if word_id is None:
                aligned_labels.append(-100)

            elif word_id != previous_word_id:
                label = labels[word_id]
                aligned_labels.append(
                    label2id[label]
                )

            else:
                aligned_labels.append(-100)

            previous_word_id = word_id

        all_labels.append(aligned_labels)

    tokenized_inputs["labels"] = all_labels

    return tokenized_inputs
```

```python
def prepare_chunked_dataset(dataset):
    chunked_dataset = []

    for item in dataset:
        result = tokenize_and_align_labels_with_chunks(
            item["tokens"],
            item["labels"]
        )

        for i in range(len(result["input_ids"])):
            chunked_dataset.append({
                "input_ids": result["input_ids"][i],
                "attention_mask": result["attention_mask"][i],
                "labels": result["labels"][i]
            })

    return chunked_dataset
```

```python
def compute_metrics(eval_pred):

    predictions, labels = eval_pred

    predictions = np.argmax(
        predictions,
        axis=2
    )

    true_predictions = []
    true_labels = []

    for prediction, label in zip(predictions, labels):

        current_predictions = []
        current_labels = []

        for predicted_id, label_id in zip(prediction, label):

            if label_id != -100:

                current_predictions.append(
                    id2label[int(predicted_id)]
                )

                current_labels.append(
                    id2label[int(label_id)]
                )

        true_predictions.append(current_predictions)
        true_labels.append(current_labels)

    precision = precision_score(
        true_labels,
        true_predictions
    )

    recall = recall_score(
        true_labels,
        true_predictions
    )

    f1 = f1_score(
        true_labels,
        true_predictions
    )

    return {
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
def tokenize_test_with_metadata(item):
    tokens = item["tokens"]
    labels = item["labels"]

    tokenized = tokenizer(
        tokens,
        is_split_into_words=True,
        truncation=True,
        max_length=512,
        stride=128,
        return_overflowing_tokens=True
    )

    chunks = []

    for chunk_index in range(len(tokenized["input_ids"])):

        word_ids = tokenized.word_ids(
            batch_index=chunk_index
        )

        chunks.append({
            "document_id": item["document_id"],
            "chunk_index": chunk_index,
            "input_ids": tokenized["input_ids"][chunk_index],
            "attention_mask": tokenized["attention_mask"][chunk_index],
            "word_ids": word_ids
        })

    return chunks
```

```python
from collections import defaultdict
import torch

word_logits = defaultdict(list)

for prediction in sample_predictions:

    word_ids = prediction["word_ids"]
    logits = prediction["logits"]

    previous_word_id = None

    for position, word_id in enumerate(word_ids):

        if word_id is None:
            continue

        # Use only the first WordPiece of each original word
        if word_id == previous_word_id:
            continue

        word_logits[word_id].append(
            logits[position]
        )

        previous_word_id = word_id
```

```python
single_count = 0
overlap_count = 0

for word_id, scores in word_logits.items():

    if len(scores) == 1:
        single_count += 1

    elif len(scores) > 1:
        overlap_count += 1

print("Unique tokenized words:", len(word_logits))
print("Words appearing once:", single_count)
print("Words appearing in overlap:", overlap_count)
```

```python
final_predicted_labels = {}

for word_id, scores in word_logits.items():

    stacked_scores = torch.stack(scores)

    average_scores = stacked_scores.mean(dim=0)

    predicted_id = torch.argmax(average_scores).item()

    final_predicted_labels[word_id] = id2label[predicted_id]
```

```python
document_predictions = []
document_true_labels = []

for word_id in sorted(final_predicted_labels.keys()):

    document_predictions.append(
        final_predicted_labels[word_id]
    )

    document_true_labels.append(
        sample_test_item["labels"][word_id]
    )

print("Predictions:", len(document_predictions))
print("True labels:", len(document_true_labels))

print("\nFirst 20 predictions:")
print(document_predictions[:20])

print("\nFirst 20 true labels:")
print(document_true_labels[:20])
```

```python
def predict_document_overlap_aware(item):

    chunks = tokenize_test_with_metadata(item)

    word_logits = defaultdict(list)

    model.eval()

    for chunk in chunks:

        input_ids = torch.tensor(
            [chunk["input_ids"]]
        ).to(model.device)

        attention_mask = torch.tensor(
            [chunk["attention_mask"]]
        ).to(model.device)

        with torch.no_grad():
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        logits = outputs.logits[0].cpu()

        previous_word_id = None

        for position, word_id in enumerate(chunk["word_ids"]):

            if word_id is None:
                continue

            if word_id == previous_word_id:
                continue

            word_logits[word_id].append(
                logits[position]
            )

            previous_word_id = word_id

    predicted_labels = []
    true_labels_document = []

    for word_id in sorted(word_logits.keys()):

        stacked_scores = torch.stack(
            word_logits[word_id]
        )

        average_scores = stacked_scores.mean(
            dim=0
        )

        predicted_id = torch.argmax(
            average_scores
        ).item()

        predicted_labels.append(
            id2label[predicted_id]
        )

        true_labels_document.append(
            item["labels"][word_id]
        )

    return predicted_labels, true_labels_document
```

```python
document_predictions_all = []
document_true_labels_all = []

for i, item in enumerate(test_ner_data):

    predictions, labels = predict_document_overlap_aware(
        item
    )

    document_predictions_all.append(
        predictions
    )

    document_true_labels_all.append(
        labels
    )

    print(
        f"Document {i + 1}/{len(test_ner_data)}:",
        len(predictions),
        "words"
    )
```

```python
from seqeval.metrics import (
    precision_score,
    recall_score,
    f1_score
)

document_precision = precision_score(
    document_true_labels_all,
    document_predictions_all
)

document_recall = recall_score(
    document_true_labels_all,
    document_predictions_all
)

document_f1 = f1_score(
    document_true_labels_all,
    document_predictions_all
)

print("Overlap-aware document-level results")
print("------------------------------------")
print(f"Precision: {document_precision:.4f}")
print(f"Recall:    {document_recall:.4f}")
print(f"F1:        {document_f1:.4f}")
```

# Final Model Development and Evaluation

## Project Objective

The objective of this experiment was to fine-tune BioClinicalBERT for
fine-grained clinical Named Entity Recognition (NER).

The model identifies clinical concepts such as diseases, symptoms,
medications, procedures, laboratory values, anatomical structures,
dosages, dates, and other clinical entities from clinical text.

The dataset contains 200 annotated clinical documents with 41 clinical
entity categories.

BIO encoding was used to represent entity boundaries:

- B = Beginning of an entity
- I = Inside an entity
- O = Outside an entity

A total of 82 BIO labels were observed in the training data.

---

## Dataset Split

Documents were split at the document level to prevent information leakage.

| Dataset | Documents | Percentage |
|---|---:|---:|
| Training | 140 | 70% |
| Validation | 30 | 15% |
| Test | 30 | 15% |
| Total | 200 | 100% |

Random seed: 42

The held-out test set was not used during model development or
hyperparameter selection.

---

## Base Model

Pretrained model:

`emilyalsentzer/Bio_ClinicalBERT`

BioClinicalBERT was selected because it provides contextual language
representations pretrained on clinical text.

A token-classification head was added for the 82 observed BIO labels and
fine-tuned using the annotated training data.

---

## Tokenization and BIO Label Alignment

BioClinicalBERT uses WordPiece tokenization, which can divide one original
word into multiple subword tokens.

For example:

Original word:

`palpitations`

WordPiece representation:

`p`, `##al`, `##pit`, `##ations`

To prevent one entity label from being counted multiple times, the original
BIO label was assigned only to the first WordPiece.

Continuation WordPieces and special tokens such as `[CLS]` and `[SEP]`
were assigned a label value of `-100`, causing them to be ignored during
loss calculation and evaluation.

---

## Long-Document Processing

Many clinical documents exceeded BERT's maximum sequence length of
512 tokens.

Instead of truncating these documents, overlapping windows were used:

- Maximum sequence length: 512
- Stride: 128

This allowed information near chunk boundaries to appear in more than
one window and prevented large portions of long clinical documents from
being discarded.

Training documents were transformed into 286 chunks and validation
documents into 64 chunks.

---

# Model Development

## Selected Training Configuration

The final model configuration was:

| Parameter | Value |
|---|---|
| Base model | BioClinicalBERT |
| Learning rate | 5e-5 |
| Epochs | 5 |
| Training batch size | 2 |
| Gradient accumulation | 2 |
| Effective batch size | 4 |
| Weight decay | 0.01 |
| Maximum sequence length | 512 |
| Stride | 128 |
| Random seed | 42 |
| Model selection metric | Validation entity-level F1 |

The best validation checkpoint occurred at Epoch 4.

---

## Validation Performance

The selected model achieved the following overlap-aware document-level
validation performance:

| Metric | Score |
|---|---:|
| Precision | 0.5505 |
| Recall | 0.6176 |
| F1 | 0.5821 |

These metrics are entity-level NER metrics rather than ordinary
classification accuracy.

---

# Controlled Experiments

Several controlled experiments were conducted before final model selection.

## Five Epoch Training

Increasing training from three to five epochs improved validation
performance.

The best validation performance occurred at Epoch 4.

Validation F1:

`0.5821`

Performance stopped improving after this point, suggesting that additional
epochs were unlikely to provide meaningful improvement under the same
configuration.

---

## Lower Learning Rate

A learning rate of `3e-5` was evaluated while keeping the remaining
training configuration similar.

The resulting validation F1 was approximately:

`0.540`

This did not outperform the `5e-5` configuration and was therefore not
selected.

---

## Class-Weighted Loss

The training data contained substantial class imbalance. Some entity
categories occurred thousands of times while others had very few examples.

A class-weighted cross-entropy experiment was therefore conducted to test
whether giving additional importance to less frequent labels would improve
performance.

Results:

| Metric | Weighted Model |
|---|---:|
| Precision | 0.5139 |
| Recall | 0.6190 |
| F1 | 0.5616 |

The weighted model slightly increased recall but reduced precision and
overall F1.

Therefore, class weighting was rejected and the standard cross-entropy
model was retained.

---

# Error Analysis

Error analysis was performed to understand the model's limitations rather
than relying only on the aggregate F1 score.

Among analyzed incorrect token-level predictions:

| Error Type | Percentage |
|---|---:|
| Wrong entity type | 49.32% |
| Entity ↔ O error | 41.19% |
| BIO boundary-only error | 9.49% |

The relatively small proportion of boundary-only errors suggested that
incorrect entity boundaries were not the dominant problem.

The largest source of error was semantic confusion between clinical entity
categories.

Examples of frequently confused categories included:

- Disease_disorder → Sign_symptom
- Diagnostic_procedure → Detailed_description
- Detailed_description → Biological_structure
- Sign_symptom → Disease_disorder
- Family_history → History

This indicates that future improvement should focus on semantic
discrimination between closely related clinical concepts rather than only
BIO boundary prediction.

# Overlap-Aware Document Reconstruction

Because long documents were processed using overlapping chunks, the same
original word could appear in multiple chunks.

During document-level evaluation:

1. Predictions were mapped from WordPiece tokens back to original words.
2. Only the first WordPiece of each original word was evaluated.
3. Logits for words appearing in multiple overlapping chunks were averaged.
4. The label with the highest averaged logit was selected.
5. Original document-level prediction sequences were reconstructed.
6. Entity-level precision, recall, and F1 were calculated using seqeval.

Tokenizer-dropped whitespace/newline tokens that produced no WordPiece were
excluded from reconstruction.

This is an evaluation limitation because some dropped tokens contained BIO
labels and their removal can affect reconstructed BIO sequences.

# Final Model Selection

The final selected approach was:

BioClinicalBERT  
→ 82 observed BIO labels  
→ WordPiece-aware label alignment  
→ 512-token overlapping windows  
→ 128-token stride  
→ 5-epoch training schedule  
→ best validation checkpoint  
→ overlap-aware document reconstruction

Final held-out performance:

**Precision: 58.05%**  
**Recall: 65.00%**  
**Entity-level Micro F1: 61.33%**

The model should be considered an experimental clinical NLP model rather
than a production clinical decision-support system.

# Key Findings

1. BioClinicalBERT successfully learned fine-grained clinical entity
   extraction across a large entity taxonomy.

2. Long clinical documents required explicit handling because many exceeded
   BERT's 512-token context limit.

3. Increasing training duration improved performance until validation
   performance plateaued.

4. Lowering the learning rate did not improve the selected model.

5. Class-weighted loss increased recall slightly but reduced precision and
   overall F1.

6. Error analysis showed that semantic entity-type confusion was a much
   larger problem than BIO boundary errors.

7. The final selected model generalized to the held-out test set with an
   entity-level micro F1 of 0.6133.

---

# Limitations and Future Work

The model still makes meaningful errors and should not be considered
production-ready for clinical decision-making.

Potential future improvements include:

- reviewing annotation consistency for frequently confused entity types
- increasing examples for underrepresented entity categories
- investigating clinically meaningful taxonomy simplification
- comparing stronger clinical language-model encoders
- improving long-document contextual modeling
- exploring span-based NER or structured decoding approaches
- evaluating performance on additional external clinical datasets

The next stage of this project focuses on integrating the NER model into a
Clinical Document Review Assistant that converts extracted clinical
entities into structured, interpretable information.
