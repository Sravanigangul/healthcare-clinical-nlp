```python
import json

with open("../data/train_ner_data.json", "r", encoding="utf-8") as f:
    train_ner_data = json.load(f)

with open("../data/val_ner_data.json", "r", encoding="utf-8") as f:
    val_ner_data = json.load(f)

with open("../data/test_ner_data.json", "r", encoding="utf-8") as f:
    test_ner_data = json.load(f)

print("Train:", len(train_ner_data))
print("Validation:", len(val_ner_data))
print("Test:", len(test_ner_data))
```

```python



all_labels = set()
for data in train_ner_data:
    all_labels.update(data["labels"])

all_labels = sorted(list(all_labels))

print("Number of unique labels:", len(all_labels))
print("Unique labels:", all_labels)
```

```python
label2id = {
    label : i
    for i, label in enumerate(all_labels)
}

label2id
```

```python
id2label = {
    i : label
    for label, i in label2id.items()
}

id2label
```
